# Student Performance Multivariate Analysis

```python
import pandas as pd
import numpy as np
from ucimlrepo import fetch_ucirepo

from missingfcup import MissingData
from missingfcup import Panel

from mdatagen.multivariate.mMCAR import mMCAR
from mdatagen.multivariate.mMAR  import mMAR
from mdatagen.multivariate.mMNAR import mMNAR
```

```python
# initial data set loading
student_performance = fetch_ucirepo(id=320)
X_raw = student_performance.data.features
y = student_performance.data.targets.iloc[:, 0].to_numpy()

# check how many missing values we have in the dataset
X_raw.isna().sum()
```

```python
# check dataset shape
X_raw.shape
```

the dataset is already complete with no missing values, so we can use it directly.

we encode all categorical columns as integer codes so that every column is numeric

```python
X = X_raw.copy()
cat_cols = X.select_dtypes(include=["object", "category"]).columns
for col in cat_cols:
    X[col] = pd.factorize(X[col])[0]
X = X.astype(float)

X.isna().sum()
```

```python
X.shape
```

now we have a clean numeric dataset to work with, we can start our analysis.

unlike the univariate analysis (which introduced missingness into a single column), here we inject missing values into multiple columns simultaneously to simulate realistic multivariate missingness patterns. 

we use `G1` (the first-period grade) as the target variable `y`, since it is the most informative predictor for the MNAR mechanism.

### MCAR generation

generate a dataset with MCAR missingness across multiple columns simultaneously.

missingness is completely random and does not depend on any observed variable or on `G1`.

```python
np.random.seed(42)
data_mcar = mMCAR(X=X, y=y, missing_rate=30).random()

data_mcar.isna().sum()
```

```python
data_mcar.shape
```

### MAR generation

generate a dataset with MAR missingness across multiple columns simultaneously.

missingness is introduced using the random method from mdatagen. in `n_xmiss=15` randomly selected column pairs, the rows with the lowest values of the observed column are made missing in the other column; so missingness depends on observed values, not on the missing value itself.


```python
np.random.seed(42)
data_mar = mMAR(X=X, y=y, n_xmiss=15).random(missing_rate=30)

data_mar.isna().sum()
```

```python
data_mar.shape
```

### MNAR generation

generate a dataset with MNAR missingness across multiple columns simultaneously.

missingness is genereated on the prediction target `G1` using `threshold=1`. students with the lowest academic performance are most likely to have their feature records missing, simulating the case where the worst-performing students are systematically underrepresented in the data.

since `G1` is the target (`y`) and is absent from the generated feature matrix, its effect must be inferred through correlated proxies such as `failures`, `absences`, and `studytime`.

```python
np.random.seed(42)
data_mnar = mMNAR(X=X, y=y, threshold=1, n_xmiss=15).random(missing_rate=30)

data_mnar.isna().sum()
```

```python
data_mnar.shape
```

```python
data_mcar = pd.read_csv("data/mcar_multi.csv")
data_mar  = pd.read_csv("data/mar_multi.csv")
data_mnar = pd.read_csv("data/mnar_multi.csv")
```

## Mechanism Discovery

we now look for structural patterns in the missingness using heatmaps, correlation matrices, dendrograms, parallel coordinates, and value-missingness correlations.

```python
panel = Panel(
    plots=[
        md_mcar.matrix(title="MCAR"),
        md_mar.matrix(title="MAR"),
        md_mnar.matrix(title="MNAR"),
    ],
    title="Row-wise Missingness Pattern",
    width=1800,
    height=600,
)

panel.show()
panel.save(f'plots/multi/{panel._download_filename}.png', save_individual=True)
```

* MCAR: Missing values are spread uniformly across all 30 columns with no visible pattern. Consistent with MCAR.

* MAR: 15 columns are entirely missing while the remaining 15 are entirely present, forming a clear striped column pattern. The same subset of rows is missing across all affected columns simultaneously, indicating a shared structural driver. We will explore further.

* MNAR: A similar column-based pattern appears, with 15 columns showing concentrated missingness and 15 fully present. The affected columns differ from MAR, reflecting a different underlying driver. We will explore further.

the correlation heatmaps can help us quantify how strongly columns miss together. the dendrogram and parallel coordinates then summarise the column-level structure.

```python
panel = Panel(
    plots=[
        md_mcar.heatmap(title="MCAR"),
        md_mar.heatmap(title="MAR", drop_constant_columns=True),
        md_mnar.heatmap(title="MNAR", drop_constant_columns=True),
    ],
    title="Column Missingness Correlation",
    width=1800,
    height=600,
)

panel.show()
panel.save(f'plots/multi/{panel._download_filename}.png', save_individual=True)
```

* MCAR: near-zero correlations throughout. no column pairs go missing together systematically. MCAR confirmed.

* MAR: strong correlations visible among the 15 affected columns, which always go missing in the same rows. this produces a dense cluster of near-perfect correlations among those columns.

* MNAR: similar strong correlations among the 15 affected columns, also always missing in the same rows. both MAR and MNAR show this block structure because their missingness is systematic, not random.

note: fewer rows and columns appear in MAR and MNAR because unaffected columns have no missing values and are dropped via `drop_constant_columns=True`.

the dendrogram groups columns by how similar their missingness patterns are. early merges mean columns miss in the same rows.

```python
plot_mcar = md_mcar.dendrogram(title="Missingness Clustering (MCAR)")
plot_mar  = md_mar.dendrogram(title="Missingness Clustering (MAR)")
plot_mnar = md_mnar.dendrogram(title="Missingness Clustering (MNAR)")

plot_mcar.show()
plot_mcar.save(f'plots/multi/{plot_mcar._download_filename}.png')

plot_mar.show()
plot_mar.save(f'plots/multi/{plot_mar._download_filename}.png')

plot_mnar.show()
plot_mnar.save(f'plots/multi/{plot_mnar._download_filename}.png')
```

* MCAR: all columns merge at distances close to 1.0, confirming that no columns go missing together.

* MAR: the 15 affected columns form a distinct cluster merging at distances of 0.5–0.65, well below the non-missing columns which all join at ~1.0. this tight grouping reflects that they are always missing in the same rows.

* MNAR: a similar cluster of affected columns merges at low distances (~0.5–0.85), separated from the non-missing columns at ~1.0. the clustering is comparably tight to MAR, not weaker.

```python
plot_mcar_pc = md_mcar.parallel_coordinates(
    title="MCAR",
    selected_columns=["school", "studytime", "failures", "absences", "health", "goout"],
    missingness_color_column="school",
)
plot_mar_pc = md_mar.parallel_coordinates(
    title="MAR",
    selected_columns=["school", "studytime", "failures", "absences", "health", "goout"],
    missingness_color_column="school",
)
plot_mnar_pc = md_mnar.parallel_coordinates(
    title="MNAR",
    selected_columns=["school", "studytime", "failures", "absences", "health", "goout"],
    missingness_color_column="school",
)

panel = Panel(
    plots=[plot_mcar_pc, plot_mar_pc, plot_mnar_pc],
    title="Parallel Coordinates by school Missingness",
    description="lines colored green = school present, red = school missing",
    width=1800,
    height=600,
)

panel.show()
panel.save(f'plots/multi/{panel._download_filename}.png', save_individual=True)

plot_mcar_pc.save(f'plots/multi/{plot_mcar_pc._download_filename}.png')
plot_mar_pc.save(f'plots/multi/{plot_mar_pc._download_filename}.png')
plot_mnar_pc.save(f'plots/multi/{plot_mnar_pc._download_filename}.png')
```

in MCAR, school-missing and school-present lines are distributed uniformly across all selected variables with no visible separation. in MAR, the school-missing lines concentrate at lower health values, confirming the health-driven MAR row selection. in MNAR, the school-missing lines (low-G1 students removed) cluster at lower studytime values, indirectly reflecting the systematic removal of academically at-risk students whose records are underrepresented in the observed data.

we will now explore a plot that shows us if the presence of a value has a direct relation in another one go missing

```python
panel = Panel(
    plots=[
        md_mcar.heatmap(kind="predictive", title="MCAR"),
        md_mar.heatmap(kind="predictive", title="MAR", drop_constant_columns=True),
        md_mnar.heatmap(kind="predictive", title="MNAR", drop_constant_columns=True),
    ],
    title="Present-vs-Missing Correlation",
    width=1800,
    height=600,
)

panel.show()
panel.save(f'plots/multi/{panel._download_filename}.png', save_individual=True)
```

* MCAR: No pattern. MCAR confirmed.

* MAR: High values here, show us what we expected, the presence of a value explains the missing in another. Strong MAR evidence. The clearest driver-partner pairs are: `health`→`school`, `Fedu`→`sex` and `famrel`, `age`→`famsup` and `activities`, `Walc`→`paid`, and `Medu`→`romantic`. 

* MNAR: Similar to MCAR, and we don't see a clear driver for missingess. We need further plots to discard MAR and confirm MNAR.

the value-missingness heatmap identified the following driver-partner pairs in MAR: `health`→`school`, `Fedu`→`sex` and `famrel`, `age`→`famsup` and `activities`, `Walc`→`paid`, and `Medu`→`romantic`. we now sort the heatmap by each driver column to confirm that missingness in its partner column forms a contiguous block.

```python
panel = Panel(
    plots=[
        md_mar.matrix(
            title="health → school",
            selected_columns=["health", "school"],
            sort_by="health", ascending=True,
        ),
        md_mar.matrix(
            title="Fedu → sex, famrel",
            selected_columns=["Fedu", "sex", "famrel"],
            sort_by="Fedu", ascending=True,
        ),
        md_mar.matrix(
            title="age → famsup, activities",
            selected_columns=["age", "famsup", "activities"],
            sort_by="age", ascending=True,
        ),
        md_mar.matrix(
            title="Walc → paid",
            selected_columns=["Walc", "paid"],
            sort_by="Walc", ascending=True,
        ),
        md_mar.matrix(
            title="Medu → romantic",
            selected_columns=["Medu", "romantic"],
            sort_by="Medu", ascending=True,
        ),
    ],
    title="MAR Driver-Partner Pairs",
    description="health, Fedu, age, Walc, Medu drive missingness in partner columns",
    max_plots=5,
    width=1800,
    height=700,
)

panel.show()
panel.save(f'plots/multi/{panel._download_filename}.png', save_individual=True)
```

four of the five sorted heatmaps confirm a MAR relationship: `health`→`school`, `Fedu`→`sex`/`famrel`, `age`→`famsup`/`activities`, and `Walc`→`paid` all produce a clear contiguous block in the partner column when sorted by the driver. `Medu`→`romantic` shows a more fragmented pattern with alternating present and missing bands rather than a single clean block, suggesting the relationship is weaker or that the injection added noise to this pair.

we now explore the MNAR scenario. we test the strongest proxy variables (`failures`, `absences`) to confirm that no observed column produces a clean contiguous block.

```python
panel = Panel(
    plots=[
        md_mnar.matrix(
            title="sorted by failures",
            sort_by="failures", ascending=False
        ),
        md_mnar.matrix(
            title="sorted by absences",
            sort_by="absences", ascending=False
        ),
    ],
    title="MNAR Pattern by Candidate Driver Columns",
    description="sorted by failures and absences to expose the unobserved G1 driver",
    width=1800,
    height=600,
)

panel.show()
panel.save(f'plots/multi/{panel._download_filename}.png', save_individual=True)
```

Sorting by `failures` and `absences` produces only weak or partial clustering, unlike the clear blocks seen in MAR. Neither variable supports a MAR explanation.

## Comparative Discovery

since missingness was artificially introduced, the original complete dataset is available as ground truth. we use density plots, boxplots, and scatterplots to compare the distributions of observed values between present and missing groups, revealing how each mechanism shapes the data differently.

to confirm MNAR, we compare the observed distributions of proxy variables between MCAR and MNAR. a distribution shift in MNAR but not MCAR is direct evidence that the at-risk cohort was selectively removed.

```python
panel = Panel(
    plots=[
        md_mcar.density(x="studytime", color_by="school", title="MCAR"),
        md_mnar.density(x="studytime", color_by="school", title="MNAR"),
    ],
    title="studytime Distribution by school Missingness",
    width=1800,
    height=600,
)

panel.show()
panel.save(f'plots/multi/{panel._download_filename}.png')
```

```python
panel = Panel(
    plots=[
        md_mcar.density(x="health", color_by="school", title="MCAR"),
        md_mnar.density(x="health", color_by="school", title="MNAR"),
    ],
    title="health Distribution by school Missingness",
    width=1800,
    height=600,
)

panel.show()
panel.save(f'plots/multi/{panel._download_filename}.png')
```

* MCAR: Both curves (`school` present vs missing) overlap closely for both `studytime` and `health`. No shift. Consistent with random missingness.

* MNAR (`studytime`): the visible range is compressed to 1–2 (compared to 1–4 in MCAR), because the MNAR mechanism also affected the `studytime` column, removing rows that had higher values. within this range, the school-missing group is sharply concentrated at `studytime=1`, while the school-present group is more spread. the compression itself is indirect evidence of the systematic removal of the at-risk cohort.

* MNAR (`health`): the school-missing group concentrates at `health=5`, while the school-present group is more spread with a notable mass around `health=3`. the distributions differ between groups, though neither `studytime` nor `health` is the true driver, they are observable correlates of the unobserved G1.

```python
panel_studytime = Panel(
    plots=[
        md_mcar.boxplot(x="studytime", color_by="school", title="MCAR: studytime by school"),
        md_mnar.boxplot(x="studytime", color_by="school", title="MNAR: studytime by school"),
    ],
    title="studytime Boxplots by school Missingness",
    description="MNAR: school-missing group shifts toward lower studytime",
    width=1800,
    height=600,
)

panel_studytime.show()
panel_studytime.save(f'plots/multi/{panel_studytime._download_filename}.png')

panel_goout = Panel(
    plots=[
        md_mcar.boxplot(x="goout", color_by="school", title="MCAR: goout by school"),
        md_mnar.boxplot(x="goout", color_by="school", title="MNAR: goout by school"),
    ],
    title="goout Boxplots by school Missingness",
    description="MCAR: no difference between groups",
    width=1800,
    height=600,
)

panel_goout.show()
panel_goout.save(f'plots/multi/{panel_goout._download_filename}.png')
```

In MCAR, the present and missing groups show nearly identical distributions for both `studytime` and `goout`. In MNAR, the entire `studytime` axis is compressed to 1–1.5 for both groups, and no clear separation is visible between school-present and school-missing. The `goout` distributions are also similar between groups. The density plots below show the distributional shift more clearly.

we now apply boxplot, density, and scatterplot to compare all three mechanisms using one representative column pair each.

```python
panel = Panel(
    plots=[
        md_mcar.boxplot(x="studytime", color_by="absences", title="MCAR: studytime by absences"),
        md_mar.boxplot(x="health", color_by="school", title="MAR: health by school"),
        md_mnar.boxplot(x="studytime", color_by="school", title="MNAR: studytime by school"),
    ],
    title="Value Distribution by Missingness Indicator (Boxplot)",
    width=1800,
    height=600,
)

panel.show()
panel.save(f'plots/multi/{panel._download_filename}.png')
```

* MCAR: The distribution of `studytime` is similar between present and missing `absences` groups. No pattern. Consistent with MCAR.

* MAR: The school-missing group spans a wider range of `health` values (Q1 ≈2.5) compared to the school-present group (Q1 ≈3). Both medians are close (~4), but the missing group has a heavier lower tail, reflecting that lower-health students are more likely to have `school` missing. Confirms MAR.

* MNAR: Both groups show `studytime` compressed to 1–1.5 with no clear separation between present and missing. The boxplot does not clearly distinguish the groups; the density plot below captures the distributional difference more precisely.

```python
panel = Panel(
    plots=[
        md_mcar.density(x="studytime", color_by="absences", title="MCAR: studytime by absences"),
        md_mar.density(x="health", color_by="school", title="MAR: health by school"),
        md_mnar.density(x="studytime", color_by="school", title="MNAR: studytime by school"),
    ],
    title="Value Distribution by Missingness Indicator (Density)",
    width=1800,
    height=600,
)

panel.show()
panel.save(f'plots/multi/{panel._download_filename}.png')
```

* MCAR: The density curves for `studytime` split by `absences` missingness overlap throughout. No pattern. MCAR confirmed.

* MAR: The missing `school` density is concentrated around `health=3`, while the present-school curve peaks at `health=5`. The clear separation confirms `health` as the MAR driver for `school` missingness.

* MNAR: The missing-school curve is sharply concentrated at `studytime=1`, while the school-present curve is broader across 1–2. The visible range is compressed to 1–2 because `studytime` is itself one of the MNAR-affected columns, with higher-studytime rows removed. The shift is indirect evidence of the unobserved G1 driver, not a direct MAR relationship. MNAR confirmed.

```python
panel = Panel(
    plots=[
        md_mcar.scatterplot(x="studytime", y="absences", title="MCAR: studytime vs absences"),
        md_mar.scatterplot(x="health", y="absences", missingness_color_column="school", title="MAR: health vs absences (by school)"),
        md_mnar.scatterplot(x="studytime", y="goout", missingness_color_column="school", title="MNAR: studytime vs goout (by school)"),
    ],
    title="Variable Relationships Colored by Missingness",
    width=1800,
    height=600,
)

panel.show()
panel.save(f'plots/multi/{panel._download_filename}.png')
```

* MCAR: Missing values in `studytime` and `absences` appear as offsets at the axes with no spatial pattern. Points are spread randomly. MCAR confirmed.

* MAR: School-missing points (red) cluster at the left side of the plot (low `health` values). Students with lower health are systematically more likely to have `school` missing. Clear spatial separation confirms the driver-partner relationship.

* MNAR: School-missing points (red) cluster at the far left of the x-axis (low `studytime`, range 1–2), reflecting the compressed range caused by the MNAR mechanism. No concentration is visible along the `goout` axis, red points span the full goout range. The low-studytime clustering is an indirect signature of the unobserved G1 driver, not a direct causal relationship. MNAR confirmed.

## Statistical Test: Little's MCAR Test

```python
results = {
    "MCAR": md_mcar.littles_mcar_test(),
    "MAR":  md_mar.littles_mcar_test(),
    "MNAR": md_mnar.littles_mcar_test(),
}

pd.DataFrame({k: v[["chi2", "df", "p_value"]] for k, v in results.items()}).T
```

* MCAR: p=0.237. We fail to reject MCAR. Consistent with all visual evidence above.

* MAR: p≈0. MCAR is decisively rejected. The structured driver-partner mechanism is confirmed.

* MNAR: p=1.0. Little's test fails here because the missing pattern is too globally consistent, a known limitation of the test. The visual and distributional evidence above confirms MNAR.

## Conclusions

* MCAR: No observed variable explains missingness in any column. Missing values are spread randomly across all rows and columns, with no structural pattern visible in any of the plots. Little's test (p=0.237) confirms MCAR.

* MAR: The value-missingness heatmap identified 7 driver-partner pairs: `health`→`school`, `Fedu`→`sex` and `famrel`, `age`→`famsup` and `activities`, `Walc`→`paid`, and `Medu`→`romantic`. Sorting the heatmap by each driver showed a clear contiguous missing block in the partner column. Boxplot, density, and scatter plots confirmed that the missing `school` group concentrates at lower `health` values. Little's test rejects MCAR (p≈0). MAR confirmed.

* MNAR: No observed variable consistently predicts missingness, the present-missing heatmap is flat and the ordered heatmaps show no clean block, ruling out MAR. Driver columns `studytime` and `goout` (both with 0 missing in MNAR) show distributional shifts between the school-present and school-missing groups. This is indirect evidence: those columns correlate with G1, the unobserved driver. The `target` column (G1) was excluded from the analysis to simulate the real-world scenario where the driver is unavailable. Little's test gives p=1.0, a known limitation for systematic missingness. MNAR confirmed.
